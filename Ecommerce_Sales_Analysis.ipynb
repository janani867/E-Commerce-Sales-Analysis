```python
import matplotlib.pyplot as plt

monthly_sales = df.groupby('month')['total_sales'].sum()
monthly_sales = monthly_sales.reindex(['January', 'February'])

plt.figure(figsize=(8, 5))

bars = plt.bar(monthly_sales.index, monthly_sales.values)

plt.title('Monthly Sales Analysis')
plt.xlabel('Month')
plt.ylabel('Total Sales')

# Show values on top of bars
for bar, value in zip(bars, monthly_sales.values):
    plt.text(
        bar.get_x() + bar.get_width()/2,
        bar.get_height(),
        f'{value/1e6:.2f}M',
        ha='center',
        va='bottom'
    )

plt.tight_layout()
plt.show()
```

```python
category_sales = (
    df.groupby('category')['total_sales']
      .sum()
      .sort_values(ascending=False)
)

plt.figure(figsize=(8, 5))

bars = plt.bar(category_sales.index, category_sales.values)

plt.title('Sales by Product Category')
plt.xlabel('Category')
plt.ylabel('Total Sales')

for bar, value in zip(bars, category_sales.values):
    plt.text(
        bar.get_x() + bar.get_width()/2,
        bar.get_height(),
        f'{value/1e6:.2f}M',
        ha='center',
        va='bottom'
    )

plt.tight_layout()
plt.show()
```

```python
subcategory_sales = (
    df.groupby('sub_category')['total_sales']
      .sum()
      .sort_values(ascending=False)
)

plt.figure(figsize=(10, 6))

bars = plt.bar(subcategory_sales.index, subcategory_sales.values)

plt.title('Sales by Sub-Category')
plt.xlabel('Sub-Category')
plt.ylabel('Total Sales')
plt.xticks(rotation=30)

for bar, value in zip(bars, subcategory_sales.values):
    plt.text(
        bar.get_x() + bar.get_width()/2,
        bar.get_height(),
        f'{value/1e6:.2f}M',
        ha='center',
        va='bottom',
        fontsize=9
    )

plt.tight_layout()
plt.show()
```

```python
payment_sales = (
    df.groupby('payment_method')['total_sales']
      .sum()
      .sort_values(ascending=False)
)

plt.figure(figsize=(8, 5))

bars = plt.bar(payment_sales.index, payment_sales.values)

plt.title('Sales by Payment Method')
plt.xlabel('Payment Method')
plt.ylabel('Total Sales')

for bar, value in zip(bars, payment_sales.values):
    plt.text(
        bar.get_x() + bar.get_width()/2,
        bar.get_height(),
        f'{value/1e6:.2f}M',
        ha='center',
        va='bottom'
    )

plt.tight_layout()
plt.show()
```

```python
discount_sales = (
    df.groupby('discount_range')['total_sales']
      .sum()
      .reindex(['0-10%', '10-20%', '20-30%'])
)

plt.figure(figsize=(8, 5))

bars = plt.bar(discount_sales.index, discount_sales.values)

plt.title('Sales by Discount Range')
plt.xlabel('Discount Range')
plt.ylabel('Total Sales')

for bar, value in zip(bars, discount_sales.values):
    plt.text(
        bar.get_x() + bar.get_width()/2,
        bar.get_height(),
        f'{value/1e6:.2f}M',
        ha='center',
        va='bottom'
    )

plt.tight_layout()
plt.show()
```

```python
top_customers = (
    df.groupby('customer_name')['total_sales']
      .sum()
      .sort_values(ascending=False)
      .head(10)
)

plt.figure(figsize=(10, 6))

bars = plt.barh(top_customers.index, top_customers.values)

plt.title('Top 10 Customers by Total Sales')
plt.xlabel('Total Sales')
plt.ylabel('Customer')

plt.gca().invert_yaxis()

for bar, value in zip(bars, top_customers.values):
    plt.text(
        bar.get_width(),
        bar.get_y() + bar.get_height()/2,
        f' {value/1e3:.1f}K',
        va='center'
    )

plt.tight_layout()
plt.show()
```

```python
state_sales = (
    df.groupby('state')['total_sales']
      .sum()
      .sort_values(ascending=False)
      .head(10)
)

plt.figure(figsize=(10, 6))

bars = plt.barh(state_sales.index, state_sales.values)

plt.title('Top 10 States by Total Sales')
plt.xlabel('Total Sales')
plt.ylabel('State')

plt.gca().invert_yaxis()

for bar, value in zip(bars, state_sales.values):
    plt.text(
        bar.get_width(),
        bar.get_y() + bar.get_height()/2,
        f' {value/1e6:.2f}M',
        va='center'
    )

plt.tight_layout()
plt.show()
```

```python
category_quantity = (
    df.groupby('category')['quantity']
      .sum()
      .sort_values(ascending=False)
)

plt.figure(figsize=(8, 5))

bars = plt.bar(category_quantity.index, category_quantity.values)

plt.title('Quantity Sold by Category')
plt.xlabel('Category')
plt.ylabel('Units Sold')

for bar, value in zip(bars, category_quantity.values):
    plt.text(
        bar.get_x() + bar.get_width()/2,
        bar.get_height(),
        f'{value:,}',
        ha='center',
        va='bottom'
    )

plt.tight_layout()
plt.show()
```
